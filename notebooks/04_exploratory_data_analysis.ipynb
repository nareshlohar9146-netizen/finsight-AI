```python
from pathlib import Path
PROJECT_ROOT = Path.cwd().parent
DATA_PATH = PROJECT_ROOT / "data" / "raw" / "application_train.csv"
df = pd.read_csv(DATA_PATH)
print(f"Dataset shape: {df.shape}")
df.head()
```

```python
target_count = df["TARGET"].value_counts().sort_index()
print(target_count)
```

```python
target_percentage = ( df["TARGET"].value_counts(normalize=True).sort_index().mul(100) )
print(target_percentage)
```

```python
plt.figure(figsize=(8,5))

sns.countplot(data=df, x="TARGET")

plt.title("Distribution of loan Payment Difficulties")
plt.xlabel("TARGET (0 = No payment difficulties, 1 = Payment difficulties)")
plt.ylabel("Number of applications")

plt.tight_layout()
plt.show()
```

```python
numeric_columns = [
    "AMT_INCOME_TOTAL",
    "AMT_CREDIT",
    "AMT_ANNUITY",
    "AMT_GOODS_PRICE",
    "DAYS_BIRTH",
    "CNT_CHILDREN",
]

df[numeric_columns].describe().T
```

```python
plt.figure(figsize=(9, 5))

sns.histplot(
    data=df,
    x="AMT_INCOME_TOTAL",
    bins=50,
    kde=True
)

plt.title("Distribution of Applicant Income")
plt.xlabel("Total Income")
plt.ylabel("Number of Applications")

plt.tight_layout()
plt.show()
```

```python
income_99percentile = df["AMT_INCOME_TOTAL"].quantile(0.99)
income_view = df[df["AMT_INCOME_TOTAL"]<=income_99percentile]

plt.figure(figsize=(9,5))

sns.histplot(
    data=income_view,
    x="AMT_INCOME_TOTAL",
    bins=50
)
plt.title("Distribution of Applicant Income - upto 99th percentile")
plt.xlabel("Total Income")
plt.ylabel("Number of applications")

plt.tight_layout()
plt.show()
```

```python
missing_report = pd.DataFrame({
    "missing_count" : df.isnull().sum(),
    "missing_percentage" : df.isnull().mean().mul(100)
})
missing_report = missing_report.sort_values(
    by="missing_percentage",
    ascending = False
)
display(missing_report.head(20).round(2))
```

```python
top_missing = missing_report.head(15).sort_values(
    "missing_percentage"
)

plt.figure(figsize = (10,7))

plt.barh(
    top_missing.index,
    top_missing["missing_percentage"]
)

plt.title("Top 15 Features by Missing Percentage")
plt.xlabel("Missing Values (%)")
plt.ylabel("Feature")

plt.tight_layout()
plt.show()
```

```python
df["AGE_YEARS"] = -df["DAYS_BIRTH"]/365.25
```

```python
plt.figure(figsize=(10, 6))

sns.histplot(
    data=df,
    x="AGE_YEARS",
    hue="TARGET",
    bins=40,
    stat="density",
    common_norm=False,
    element="step",
    fill=False
)

plt.title("Applicant Age Distribution by Payment-Difficulty Status")
plt.xlabel("Age (Years)")
plt.ylabel("Density")

plt.tight_layout()
plt.show()
```

```python
age_summary = (
    df.groupby("TARGET")["AGE_YEARS"].agg(["count","mean","median","std"])
) 
print(age_summary.round(2))
```

```python
income_summary = (
    df.groupby("TARGET")["AMT_INCOME_TOTAL"]
    .agg(["count", "mean", "median"])
)

display(income_summary.round(2))
```

```python
df["INCOME_GROUP"] = pd.qcut(
    df["AMT_INCOME_TOTAL"],
    q=5,
    duplicates="drop"
)
```

```python
df.head(20)
```

```python
income_risk = (
    df.groupby("INCOME_GROUP", observed=True)
    .agg(
        applications=("TARGET", "size"),
        payment_difficulties=("TARGET", "sum"),
        payment_difficulty_rate=("TARGET", "mean")
    )
)

income_risk["payment_difficulty_rate"] *= 100

display(income_risk.round(2))
```

```python
education_risk = (
    df.groupby("NAME_EDUCATION_TYPE")
    .agg(
        applications=("TARGET", "size"),
        payment_difficulties=("TARGET", "sum"),
        payment_difficulty_rate=("TARGET", "mean")
    )
)

education_risk["payment_difficulty_rate"] *= 100

education_risk = education_risk.sort_values(
    "payment_difficulty_rate",
    ascending=False
)

display(education_risk.round(2))
```

```python
plt.figure(figsize=(11, 6))

sns.barplot(
    data=education_risk.reset_index(),
    x="NAME_EDUCATION_TYPE",
    y="payment_difficulty_rate"
)

plt.title("Observed Payment-Difficulty Rate by Education Category")
plt.xlabel("Education Category")
plt.ylabel("Payment-Difficulty Rate (%)")

plt.xticks(rotation=35, ha="right")
plt.tight_layout()
plt.show()
```
